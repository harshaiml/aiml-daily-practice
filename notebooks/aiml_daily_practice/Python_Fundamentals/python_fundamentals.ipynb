### Word Search

```python
data = True
line = 1
word = "Python"
with open("sample.txt", "r") as f:
    while data:
        data = f.readline()
        if(word in data):
            print(f"{word} found at line {line}")
            break
        line += 1
```

```python
try:
    x = int(input("Enter x: "))
    ans = 10/x

except ZeroDivisionError:
    print(f"Divide by 0 is not allowed")

except ValueError:
    print(f"Invalid input")

else:
    print(f"ans = {ans}")

finally:
    print("Ennd of program")
```

```python
squares = []

for i in range(6):
    squares.append(i*i)

print(squares)
```

```python
sq = [i*i for i in range(6)]
print(sq)
```

```python
sq = [i*i for i in range(6) if i%2 != 0]
print(sq)
```

```python
nums = [-2, -3, 3, 4, -1, 7]

nums = [0 if val < 0 else val for val in nums]
print(nums)
```
