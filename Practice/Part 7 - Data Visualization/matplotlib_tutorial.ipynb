## Matplotlib

```python
import matplotlib.pyplot as plt
```

```python
X = [1, 2, 3, 4]
Y = [5, 6, 7, 8]

plt.plot(X, Y) # Line plot
plt.show()
```

```python
# Real data

# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The king's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

# Plot
plt.plot(years,oscar_revenue)
plt.title("Oscar Movies Revenue in each Year")
plt.xlabel("Years")
plt.ylabel("Revenue (in ($M))")
```

```python
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire","Avtar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275] # in $M

plt.plot(years, oscar_revenue, label="Oscar Movies")
plt.plot(years, non_oscar_revenue, label="Non-Oscar Movies")

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")

plt.legend()
plt.show()
```

```python
# Format Strings
# Non-Oscar Winner Movies

non_oscar_movies = ["Slumdog Millionaire","Avtar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275] # in $M

plt.plot(years, oscar_revenue, "o--r", label="Oscar Movies")
plt.plot(years, non_oscar_revenue,"<-g", label="Non-Oscar Movies")

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")

plt.legend()
plt.show()
```

```python
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire","Avtar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275] # in $M

plt.plot(years, oscar_revenue, color="aqua", marker="o", label="Oscar Movies")
plt.plot(years, non_oscar_revenue, color="#CBC3E3", marker="<",label="Non-Oscar Movies")

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")

plt.legend()
plt.show()
```

```python
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire","Avtar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275] # in $M

plt.plot(years, oscar_revenue, color="aqua", linestyle="-.", marker="o", label="Oscar Movies")
plt.plot(years, non_oscar_revenue, color="#CBC3E3", linewidth=2, linestyle="--", marker="<",label="Non-Oscar Movies")

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")

plt.legend()
plt.show()
```
