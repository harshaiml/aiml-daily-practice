## Data Visualzation

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Histograms

```python
np.random.seed(0)

scores = np.random.normal(70, 10, 100) # Normal distribution

plt.hist(scores, color="skyblue", edgecolor="black", bins=5)
plt.xlabel("Score")
plt.ylabel("# of students")
plt.show()
```

```python
np.random.seed(0)

scores = np.random.normal(70, 10, 100) # Normal distribution
bins = [0, 30, 70, 90, 100]

plt.hist(scores, color="skyblue", edgecolor="black", bins=bins)
plt.xlabel("Score")
plt.ylabel("# of students")
plt.show()
```

```python
marks = [
    25, 38, 40, 55, 61, 67, 72, 78, 83, 88,
    91, 95, 47, 53, 59, 64, 69, 74, 79, 85,
    92, 97, 33, 45, 58, 62, 71, 76, 81, 89
]

plt.hist(marks, color="skyblue", edgecolor="black",bins=bins)
plt.title("Distribution of marks for students")
plt.xlabel("Marks")
plt.ylabel("Number of students")
```

### <u>Histograms</u>

Multiple Datasets

```python
# Multiple datasets
legit_transactions = [
    2.99, 5.49, 8.99, 12.50, 14.99, 19.99, 23.45, 29.99, 34.99, 39.50,
    45.00, 49.99, 55.25, 60.00, 75.99, 89.99, 120.50, 150.00, 199.99,
    249.99, 300.75, 450.00, 600.00, 850.00, 1200.00
]

fraud_transactions = [
    50, 100, 150, 200, 300, 500, 500, 750, 1000, 1000,
    1200, 1500, 1500, 2000, 2500, 3000, 3000
]

# Transactions Amount Distrbution - Legit vs Fraud
plt.hist(fraud_transactions, color="red", bins=10, label="Fraud", edgecolor="black", alpha=0.5)
plt.hist(legit_transactions, color="green", bins=10, label="Legit", edgecolor="black", alpha=0.5)

plt.title("Transactions Amount Distrbution - Legit vs Fraud")
plt.xlabel("Amount (in $)")
plt.ylabel("Frequency")
plt.grid(True)
plt.legend()
plt.show()
```

### axvline in Histogram

```python
marks = [
    25, 38, 40, 55, 61, 67, 72, 78, 83, 88,
    91, 95, 47, 53, 59, 64, 69, 74, 79, 85,
    92, 97, 33, 45, 58, 62, 71, 76, 81, 89
]

plt.axvline(
    x=33,
    label="Passing marks (33)",
    color="green",
    linewidth="1",
    linestyle="--"
)

plt.hist(marks, color="skyblue", edgecolor="black",bins=bins)
plt.title("Distribution of marks for students")
plt.xlabel("Marks")
plt.ylabel("Number of students")
plt.show()
```

### <u>Box Plots</u>

```python
# Box Plot
data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data)
plt.grid(True)
plt.show()
```

### <u>Box Plots</u>
Additional Operations
- Multiple Datasets
- Horizontal Box Plot
- Show Mean
- Change Whisker

```python
# Multiple datasets
group1 = np.random.normal(50, 10, 100)
group2 = np.random.normal(60, 15, 100)

plt.boxplot([group1, group2], tick_labels=["group1", "group2"])
plt.grid(True)
plt.show()
```

```python
# Box Plot
data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data, vert=False)
plt.grid(True)
plt.show()
```

```python
# Box Plot
data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data, vert=False, showmeans=True)
plt.grid(True)
plt.show()
```

```python
# Multiple datasets
group1 = np.random.normal(50, 10, 100)
group2 = np.random.normal(60, 15, 100)

plt.boxplot([group1, group2], tick_labels=["group1", "group2"], showmeans="True")
plt.grid(True)
plt.show()
```

```python
# Box Plot
data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data, vert=False, showmeans=True, whis=2.0)
plt.grid(True)
plt.show()
```

### <u>Stack Plots</u>

```python
# Stack Plot
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
direct = [50, 60, 70, 80, 90, 100, 110]
organic = [30, 40, 50, 55, 60, 70, 80]
social = [20, 25, 30, 35, 40, 50, 60]

# Single day plot
# plt.pie([direct[0], organic[0], social[0]], labels=["direct", "organic", "social"], autopct="%1.1f%%")

# For multiple days
plt.stackplot(days, direct, organic, social, labels=['Direct', 'Organic', 'Social'])
plt.legend(loc='upper left')
plt.title("Website Traffic Over a Week")
plt.xlabel("Days")
plt.ylabel("Number of Visitors")
plt.show()
```
