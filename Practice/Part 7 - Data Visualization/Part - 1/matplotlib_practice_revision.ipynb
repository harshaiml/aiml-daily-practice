```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [10,20,15,25,30]

plt.plot(x, y)
plt.title("Basics Chart")
plt.xlabel("xlabel")
plt.ylabel("ylabel")
plt.show()
```

```python
years = [2019, 2020, 2021, 2022, 2023]
student_A_marks = [72, 78, 81, 87, 91]
student_B_marks = [68, 75, 79, 84, 89]

plt.plot(years, student_A_marks, marker="o", linestyle="-", label="Student A")
plt.plot(years, student_B_marks, marker="s", linestyle="--", label="Student B")

plt.title("Student A vs Student B")
plt.xlabel("Years")
plt.ylabel("Marks")
plt.legend()
plt.show()
```

```python
subjects = ["Python", "SQL", "Pandas", "ML"]
marks = [85, 90, 78, 88]

plt.bar(subjects, marks, color="#CBC3E3")

plt.title("Students Subjects Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
for i in range(len(subjects)):
    plt.text(subjects[i], marks[i]+2, str(marks[i]), ha="center")
plt.ylim(0, max(marks)+10)
plt.show()
```

```python
import numpy as np
subjects = ["Python", "SQL", "Pandas", "ML"]
student_A_marks = [85, 90, 78, 88]
student_B_marks = [80, 87, 82, 91]

x = np.arange(len(subjects))
width = 0.25

plt.bar(x-width/2,student_A_marks, width, label="Studen-A")
plt.bar(x+width/2,student_B_marks, width, label="Studen-B")

plt.title("Student A vs Student B Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.ylim(0, max(marks)+25)
plt.xticks(x, subjects)
plt.legend()
plt.show()
```

```python
import numpy as np
subjects = ["Python", "SQL", "Pandas", "ML"]
student_A_marks = [85, 90, 78, 88]
student_B_marks = [80, 87, 82, 91]
student_C_marks = [78, 68, 75, 86]

x = np.arange(len(subjects))
width = 0.25

plt.bar(x-width,student_A_marks, width, label="Studen-A")
plt.bar(x,student_B_marks, width, label="Studen-B")
plt.bar(x+width,student_C_marks, width, label="Studen-C")

plt.title("Student A vs Student B vs Student C Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.ylim(0, max(student_A_marks + student_B_marks + student_C_marks) + 10)
plt.xticks(x, subjects)
plt.legend()
plt.show()
```

```python
# Scatter Plot
names = [
    "Person A", "Person B", "Person C", "Person D", "Person E",
    "Person F", "Person G", "Person H", "Person I", "Person J"
]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
blood_pressure = [110, 115, 120, 122, 130, 128, 135, 140, 145, 150]

colors = [
    "red", "blue", "green", "orange", "purple",
    "brown", "pink", "gray", "olive", "cyan"
]

plt.scatter(age, blood_pressure, color=colors)

plt.title("Age vs Blood Pressure")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
for i in range(len(names)):
    plt.text(age[i], blood_pressure[i] + 1, names[i], ha="center")
plt.show()
```

```python
expenses = ["Food", "Rent", "Travel", "Shopping", "Savings"]
amounts = [15000, 8000, 3000, 5000, 12000]

explode = [0, 0, 0, 0, 0.1]

plt.pie(amounts, labels=expenses, autopct="%1.1f%%",
        wedgeprops= {
            "edgecolor":"black",
            "linewidth": 0.5,
            "linestyle": "--"
        },
        explode=explode,
        startangle=90,
        shadow=True,
        )
plt.title("Monthly Expenses Distribution")
plt.show()
```
