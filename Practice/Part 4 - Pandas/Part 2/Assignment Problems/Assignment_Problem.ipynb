Q1. Start by loading the IRIS dataset.

A. Display the following information
- First 10 rowa
- Shape & Data types
- Summary statistics (mean, std, min, max)

B. Select only those rows where:
- petal_lenght > 4.5
- species = "Iris-virginica"

C. Group by Species and compute:
- Average sepal_lenght
- Maximum petal_with
- Standard deviation

D. Create a new column: **"petal_ratio" = petal_lenght / petal_width**
- For each species, find the avrage petal_ratio.

```python
# Load data
df = pd.read_csv("IRIS.csv")

# Print first 10 rows
df.head(10)

# Print Shape & Data types
df.shape
df.dtypes

# Summary statistics (mean, std, min, max)
df.describe()
```

```python
# petal_lenght > 4.5
# species = "Iris-virginica"
cleand_data = df[ (df["petal_length"] > 4.5) & (df["species"] == "Iris-virginica")] 

cleand_data.head()
```

```python
# Average sepal_lenght
print(df.groupby("species")["sepal_length"].mean())
# Maximum petal_with
print(df.groupby("species")["petal_width"].max())
# Standard deviation
print(df.groupby("species")["sepal_width"].std())
```

```python
# For each species, find the avrage petal_ratio.
df['petal_ratio'] = df['petal_length'] / df['petal_width']
print(df.groupby('species')['petal_ratio'].mean())
```

Q2. Start by loading the **TITANIC** dataset.

A. Display only columns: Name, Sex, Age, Fare, Survived

B. Select passengers who:
- Are female
- Fare > 30

C. Group by Pclass and compute:
- Survival rate
- Average Fare
- Average Age

```python
# Load data
df = pd.read_csv("Titanic-Dataset.csv")
# Step A
df_selected = df[['Name', 'Sex', 'Age', 'Fare', 'Survived']]
print(df_selected)
# Step B
filterd_df = df_selected[(df_selected["Sex"] == "female") & (df_selected["Fare"] > 30)]
print(filterd_df)

# Step C
print("Survival Rate:\n", df.groupby('Pclass')['Survived'].mean())
print("Average Fare:\n", df.groupby('Pclass')['Fare'].mean())
print("Average Age:\n", df.groupby('Pclass')['Age'].mean())
```
